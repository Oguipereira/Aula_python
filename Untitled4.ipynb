```python

import numpy as np
import matplotlib.pyplot as plt

# entrada de dados
a = float(input("Digite o valor de a: "))
b = float(input("Digite o valor de b: "))
c = float(input("Digite o valor de c: "))

if a == 0:
    print("Isso não é uma equação do segundo grau")

else:
    delta = (b**2) - (4*a*c)

    if delta < 0:
        print("A equação não possui raízes reais")

    else:
        print("O valor de delta é:", delta)

        raiz1 = (-b + np.sqrt(delta)) / (2*a)
        raiz2 = (-b - np.sqrt(delta)) / (2*a)

        print("A primeira raiz é:", int(raiz1))
        print("A segunda raiz é:", int(raiz2))

# fazendo o grafico

x = np.linspace(raiz1 - 5, raiz2 + 5, 100)

y = a*x**2 + b*x + c

plt.plot(x, y)
plt.axhline(0, color='black')
plt.axvline(0, color='black')

plt.scatter([raiz1, raiz2], [0, 0], color='red')
plt.title("Gráfico da função do segundo grau")
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.grid()

plt.show()
```
