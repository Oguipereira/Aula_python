```python
import math

def area_circulo (raio):
  return math.pi*raio**2
def area_quadrado (lado):
  return lado**2
def imprimir_area(opcao_escolhida, area):
  print (f"A área da {opcao_escolhida} é {area}")


# programa principal
print ("Escolha uma figura geometrica: ")
print ("Circulo")
print ("Quadrado")
opcao_escolhida = input("Eu escolho: ")

if opcao_escolhida == "Circulo":
  raio = float(input("O valor do raio é: "))
  imprimir_area(opcao_escolhida, area_circulo(raio))

elif opcao_escolhida == "Quadrado":
  lado = float(input("O valor do lado é: "))
  imprimir_area(opcao_escolhida, area_quadrado(lado))

else:
  print("Opção invalida !")
```
